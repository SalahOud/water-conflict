```python
import ipywidgets as widgets
from IPython.display import display, Markdown

# --- Widgets ---
recherche = widgets.Text(
    value="",
    placeholder="Tapez : Nil, Indus, Mékong, Colorado, Europe...",
    description="🔎 Recherche :",
    layout=widgets.Layout(width="500px")
)

selecteur = widgets.Dropdown(
    options=df["nom"].tolist(),
    description="Zone :",
    layout=widgets.Layout(width="500px")
)

bouton = widgets.Button(
    description="Afficher la fiche",
    button_style="primary"
)

sortie = widgets.Output()

# --- Filtrage ---
def filtrer(recherche_texte):
    if not recherche_texte:
        return df
    r = recherche_texte.lower()
    mask = (
        df["nom"].str.lower().str.contains(r, na=False)
        | df["region"].str.lower().str.contains(r, na=False)
        | df["pays_concernes"].str.lower().str.contains(r, na=False)
        | df["ressource_en_eau"].str.lower().str.contains(r, na=False)
    )
    return df[mask]

def maj_liste(change=None):
    sous_df = filtrer(recherche.value)
    selecteur.options = sous_df["nom"].tolist() if len(sous_df) else ["— aucune zone —"]

recherche.observe(maj_liste, names="value")

# --- Fiche détaillée ---
def afficher_fiche(b):
    with sortie:
        sortie.clear_output()
        if selecteur.value == "— aucune zone —":
            print("Aucune zone ne correspond à votre recherche.")
            return
        ligne = df[df["nom"] == selecteur.value].iloc[0]

        md = f"""
### 💧 {ligne['nom']}

**🌍 Région :** {ligne['region']}
**👥 Pays concernés :** {ligne['pays_concernes']}
**🌊 Ressource en eau :** {ligne['ressource_en_eau']}
**📌 Type :** {ligne['type']}
**📍 Coordonnées :** {ligne['latitude']}°, {ligne['longitude']}°

---

**🚨 Statut :** {ligne['statut']}

**⚠️ Enjeux principaux :**
{ligne['enjeux']}

**📖 Description :**
{ligne['description']}

**📚 Sources :** {ligne['sources']}
"""
        display(Markdown(md))

bouton.on_click(afficher_fiche)

# --- Affichage ---
display(recherche)
display(selecteur)
display(bouton)
display(sortie)
```

```python
import torch
from transformers import AutoTokenizer, AutoModelForCausalLM

MODEL_NAME = "HuggingFaceTB/SmolLM2-1.7B-Instruct"

print("⏳ Chargement du modèle (peut prendre 1-2 min)...")

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
if tokenizer.pad_token is None:
    tokenizer.pad_token = tokenizer.eos_token

model = AutoModelForCausalLM.from_pretrained(
    MODEL_NAME,
    torch_dtype=torch.float16 if device == "cuda" else torch.float32,
).to(device)
model.eval()

print("✅ Modèle chargé :", MODEL_NAME)
print("Device :", device)
```

```python
import re
import unicodedata
from difflib import SequenceMatcher


# ================= NORMALISATION =================
def normaliser(texte):
    if texte is None:
        return ""
    texte = str(texte).lower()
    texte = unicodedata.normalize("NFD", texte)
    texte = "".join(c for c in texte if unicodedata.category(c) != "Mn")
    texte = re.sub(r"[\-/_,;.()\[\]]", " ", texte)
    texte = re.sub(r"\s+", " ", texte).strip()
    return texte


def similitude(a, b):
    return SequenceMatcher(None, a, b).ratio()


# ================= DÉTECTION INTENTION =================
def detecter_intention(question):
    q = normaliser(question)

    for region in ["afrique", "asie", "europe", "amerique", "moyen orient"]:
        if region in q and any(m in q for m in ["zone", "liste", "pays", "quels", "montre", "affiche"]):
            return ("region", region)

    if any(m in q for m in ["critique", "urgent", "grave", "pire"]):
        return ("niveau", "🔴")

    return ("rag", None)


# ================= RÉCUPÉRATION CONTEXTE =================
def recuperer_zones_pertinentes(question, top_k=5, seuil=0.72):
    """Retourne les zones les plus pertinentes + toutes si question globale."""
    q = normaliser(question)
    mots = [m for m in q.split() if len(m) >= 3]

    scores = []
    for _, ligne in df.iterrows():
        score = 0
        for champ in [ligne["nom"], ligne["ressource_en_eau"], ligne["region"], ligne["pays_concernes"], ligne["enjeux"]]:
            for mot in normaliser(champ).split():
                if len(mot) < 3:
                    continue
                if mot in mots:
                    score += 3
                else:
                    for mq in mots:
                        if similitude(mot, mq) >= seuil:
                            score += 2
                            break
        scores.append((score, ligne))

    scores.sort(key=lambda x: x[0], reverse=True)

    # Si aucun score > 0, on renvoie un échantillon large (contexte général)
    if scores[0][0] == 0:
        return df.head(10)

    return pd.DataFrame([s[1] for s in scores[:top_k] if s[0] > 0])


def construire_contexte(sous_df):
    blocs = []
    for _, z in sous_df.iterrows():
        blocs.append(f"""ZONE : {z['nom']}
Région : {z['region']}
Pays concernés : {z['pays_concernes']}
Ressource : {z['ressource_en_eau']} ({z['type']})
Statut : {z['statut']}
Enjeux : {z['enjeux']}
Description : {z['description']}
Sources : {z['sources']}""")
    return "\n\n---\n\n".join(blocs)


# ================= PROMPT SYSTÈME STRICT =================
SYSTEM_PROMPT = """Tu es WATER CONFLICT, un assistant expert spécialisé UNIQUEMENT sur :
- les conflits, tensions, crises et guerres liés aux ressources en eau dans le monde ;
- les fleuves, lacs, aquifères, barrages et bassins transfrontaliers ;
- les acteurs (États, ONU, organisations régionales) et les enjeux géopolitiques.

RÈGLES ABSOLUES :
1. Tu réponds TOUJOURS en français, de façon claire, factuelle et concise.
2. Tu te bases UNIQUEMENT sur le CONTEXTE fourni. Si une information n'y est pas, dis-le honnêtement au lieu d'inventer.
3. Si la question sort du thème (eau/conflits/géopolitique de l'eau), recentre poliment la réponse sur la thématique.
4. Ne jamais inventer de chiffres, de dates ou d'événements.
5. Structure ta réponse : réponse directe → éléments clés → sources.
6. Reste neutre politiquement, factuel, académique."""


# ================= RÉPONSE =================
def repondre(question):
    intention, cible = detecter_intention(question)

    # ---- LISTE PAR RÉGION ----
    if intention == "region":
        noms = {"afrique": "Afrique", "asie": "Asie", "europe": "Europe",
                "amerique": "Amériques", "moyen orient": "Moyen-Orient"}
        region = noms.get(cible, cible)
        sous = df[df["region"].str.lower().str.contains(cible, na=False)]
        if len(sous) == 0:
            return f"❌ Aucune zone enregistrée pour la région : **{region}**.", None
        lignes = [f"- **{r['nom']}** ({r['statut']}) — {r['pays_concernes']}" for _, r in sous.iterrows()]
        return f"### 🌍 Zones de tension en {region} ({len(sous)})\n\n" + "\n".join(lignes), None

    # ---- NIVEAU CRITIQUE ----
    if intention == "niveau":
        sous = df[df["statut"].str.startswith(cible)]
        lignes = [f"- **{r['nom']}** — {r['pays_concernes']}" for _, r in sous.iterrows()]
        return f"### 🚨 Zones critiques ({len(sous)})\n\n" + "\n".join(lignes), None

    # ---- RAG : LLM + contexte ----
    sous_df = recuperer_zones_pertinentes(question, top_k=5)
    contexte = construire_contexte(sous_df)

    messages = [
        {"role": "system", "content": SYSTEM_PROMPT},
        {"role": "user", "content": f"""CONTEXTE (base de données WATER CONFLICT) :
{contexte}

QUESTION DE L'UTILISATEUR :
{question}

Réponds en français en respectant strictement les règles."""}
    ]

    prompt = tokenizer.apply_chat_template(messages, tokenize=False, add_generation_prompt=True)
    inputs = tokenizer(prompt, return_tensors="pt", truncation=True, max_length=2048).to(device)

    with torch.no_grad():
        outputs = model.generate(
            **inputs,
            max_new_tokens=350,
            temperature=0.4,
            top_p=0.9,
            do_sample=True,
            repetition_penalty=1.25,
            no_repeat_ngram_size=4,
            pad_token_id=tokenizer.eos_token_id,
            eos_token_id=tokenizer.eos_token_id,
        )

    nouveaux = outputs[0][inputs["input_ids"].shape[1]:]
    reponse = tokenizer.decode(nouveaux, skip_special_tokens=True).strip()

    # Zones utilisées comme sources
    zones_src = ", ".join(sous_df["nom"].tolist()[:5])
    reponse += f"\n\n---\n📚 **Zones de la base mobilisées :** {zones_src}"

    return reponse, sous_df.iloc[0] if len(sous_df) else None
```

```python
import ipywidgets as widgets
from IPython.display import display, Markdown, clear_output

champ = widgets.Text(
    placeholder="Ex : Quels sont les enjeux autour du Nil ? / zones critiques en Afrique / pourquoi des tensions au Moyen-Orient ?",
    description="💬 :",
    layout=widgets.Layout(width="800px")
)
btn = widgets.Button(description="Envoyer", button_style="success")
out = widgets.Output()

def envoyer(b):
    with out:
        clear_output()
        q = champ.value.strip()
        if not q:
            print("⚠️ Posez une question.")
            return
        display(Markdown(f"**❓ {q}**\n\n⏳ _Analyse en cours..._"))
        rep, _ = repondre(q)
        clear_output()
        display(Markdown(f"**❓ {q}**\n\n---\n\n{rep}"))

btn.on_click(envoyer)

print("=" * 70)
print("🌍 WATER CONFLICT — Assistant intelligent sur les conflits de l'eau")
print("=" * 70)
print("\n💡 Exemples de questions :")
print("  • Quels sont les enjeux autour du Nil ?")
print("  • Zones critiques en Afrique ?")
print("  • Pourquoi des tensions au Moyen-Orient ?")
print("  • Quels pays sont concernés par l'Indus ?")
print("  • Explique-moi la situation du Mékong.\n")
display(champ, btn, out)
```
